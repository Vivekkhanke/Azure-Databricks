```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Transformations").getOrCreate()

data = [
    (101,"Vivek","IT",50000),
    (102,"Rahul","HR",40000),
    (103,"Amit","IT",60000),
    (104,"Sneha","Finance",70000),
    (105,"Priya","HR",45000),
    (106,"Vivek","IT",50000)
]

columns=["EmpID","Name","Dept","Salary"]

df = spark.createDataFrame(data,columns)

df.display()
```

```python
df.select("EmpID", "Salary").display()
```

```python
df.filter(df.Salary > 40000).display()
```

```python
df.select("Name","Salary").filter(df.Salary > 45000).display()
```

```python
df.where(df.Dept == 'IT').display()
```

```python
from pyspark.sql.functions import col

df.withColumn("Bonus", col("Salary") * 0.10).display()
```

```python
df.drop("Dept").display()
```

```python
df.select("Dept").distinct().display()
```

```python
df.select("Name", "Dept").orderBy("Dept").display()
```

```python
df.select("Name", "Dept").dropDuplicates().display()
```

```python
# df.select("Name","Salary").orderBy("Salary").display() #ASC
from pyspark.sql.functions import desc
df.select("Name","Salary").orderBy(desc("Salary")).display()
```
