```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Pyspark operators and Joins").getOrCreate()

# Employee Data

emp_data = [
    (101,"John", 10, 50000),
    (102, "Alice", 20, 60000),
    (103, "Bob", 30, 45000),
    (104, "David", 20, 70000),
    (105, "Emma", None, 55000)
]

emp_df =spark.createDataFrame(emp_data, ["emp_id", "emp_name", "dept_id", "salary"])

# Department Data

dept_data = [
    (10, "Sales"),
    (20, "Marketing"),
    (30, "HR")
]

dept_df = spark.createDataFrame(dept_data, ["dept_id", "dept_name"])

emp_df.display()
dept_df.display()
```

### Comparison Operators

**Employees with salary greater than 55000**

```python
emp_df.filter(emp_df.salary > 55000).display()

# Employees with salary less than or equal to 55000
emp_df.filter(emp_df.salary <= 55000).display()
```

**Department ID equals 20**

```python
emp_df.filter(emp_df.dept_id == 20).show()
```

**Logical Operators
AND (&) OR (|)**

```python
emp_df.filter((emp_df.salary > 50000) & (emp_df.dept_id == 20)).display()
```

```python
emp_df.filter(
    (emp_df.salary > 65000) | (emp_df.dept_id == 10)
).show()
```

**NOT (~)**

```python
emp_df.filter(
    ~(emp_df.dept_id == 20 )
).show()
```

#### Arithmetic Operators
**Increase salary by 10%**

```python
from pyspark.sql.functions import col

emp_df.withColumn("new_salary", col("salary") * 1.10).display()
```

### String Operators
**Employee Name Starts With A**

```python
emp_df.filter(emp_df.emp_name.startswith("A")).show()

# Employee Name ends With e
emp_df.filter(emp_df.emp_name.endswith("e")).show()
```

**Employee Name Contains 'o'**

```python
emp_df.filter(emp_df.emp_name.contains("o")).display()
```

## NULL Operators

**Find NULL Department IDs**

```python
emp_df.filter(emp_df.dept_id.isNull()).show()
```

**Find NOT NULL Department IDs**

```python
emp_df.filter(emp_df.dept_id.isNotNull()).display()
```
