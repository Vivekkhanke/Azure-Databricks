```python
employees_data = [
    (1, "Alice", 10),
    (2, "Bob", 20),
    (3, "Charlie", 30),
    (4, "Diana", 40),
    (5, "Evan", None)
]

departments_data = [
    (10, "HR"),
    (20, "Finance"),
    (30, "Engineering"),
    (50, "Marketing")
]

employees_df = spark.createDataFrame(
    employees_data,
    ["emp_id", "emp_name", "dept_id"]
)

departments_df = spark.createDataFrame(
    departments_data,
    ["dept_id", "dept_name"]
)

print("Employees DataFrame")
display(employees_df)

print("Departments DataFrame")
display(departments_df)

inner_join_df = employees_df.join(departments_df, on="dept_id", how="inner")
left_join_df = employees_df.join(departments_df, on="dept_id", how="left")
right_join_df = employees_df.join(departments_df, on="dept_id", how="right")
full_join_df = employees_df.join(departments_df, on="dept_id", how="full")

left_semi_join_df = employees_df.join(departments_df, on="dept_id", how="left_semi")
left_anti_join_df = employees_df.join(departments_df, on="dept_id", how="left_anti")

cross_join_df = employees_df.crossJoin(departments_df)

hierarchy_data = [
    (1, "Alice", None),
    (2, "Bob", 1),
    (3, "Charlie", 1),
    (4, "Diana", 2),
    (5, "Evan", 2)
]

hierarchy_df = spark.createDataFrame(
    hierarchy_data,
    ["emp_id", "emp_name", "manager_id"]
)

employee_alias = hierarchy_df.alias("e")
manager_alias = hierarchy_df.alias("m")

self_join_df = employee_alias.join(
    manager_alias,
    employee_alias.manager_id == manager_alias.emp_id,
    "left"
).select(
    employee_alias.emp_id.alias("emp_id"),
    employee_alias.emp_name.alias("employee_name"),
    manager_alias.emp_name.alias("manager_name")
)

print("Inner Join")
display(inner_join_df)

print("Left Join")
display(left_join_df)

print("Right Join")
display(right_join_df)

print("Full Join")
display(full_join_df)

print("Left Semi Join")
display(left_semi_join_df)

print("Left Anti Join")
display(left_anti_join_df)

print("Cross Join")
display(cross_join_df)

print("Self Join")
display(self_join_df)
```
