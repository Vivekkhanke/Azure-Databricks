```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Pyspark operators and Joins").getOrCreate()

# Employee Data

emp_data = [
    (101,"John", 10, 50000),
    (102, "Alice", 20, 60000),
    (103, "Bob", 30, 45000),
    (104, "David", 20, 70000),
    (105, "Emma", None, 55000)
]

emp_df =spark.createDataFrame(emp_data, ["emp_id", "emp_name", "dept_id", "salary"])

# Department Data

dept_data = [
    (10, "Sales"),
    (20, "Marketing"),
    (30, "HR")
]

dept_df = spark.createDataFrame(dept_data, ["dept_id", "dept_name"])

emp_df.display()
dept_df.display()
```

## Joins in PySpark
**Inner Join**

Returns matching records from both tables.

```python
inner_join = emp_df.join(dept_df,
                         emp_df.dept_id == dept_df.dept_id,
                         "inner")
inner_join.display()
```

## **Left Join**

```python
left_Join = emp_df.join(dept_df, emp_df.dept_id == dept_df.dept_id, "left")
left_Join.display()
```

## **Right Join**

```python
right_join = emp_df.join(dept_df, emp_df.dept_id == dept_df.dept_id, "right")
display(right_join)
```

## **Full Outer Join**

```python
full_join= emp_df.join(dept_df, emp_df.dept_id == dept_df.dept_id, "full")
display(full_join)
```

## **Left Anti Join**
**_Fetch all unmatched records from only Left table_**

```python
left_anti = emp_df.join(dept_df, emp_df.dept_id == dept_df.dept_id, "left_anti")
display(left_anti)
```

## **Left semi Join**
**_Fetch all matched records from only Left table_**

```python
left_semi = emp_df.join(dept_df, emp_df.dept_id == dept_df.dept_id, "left_semi")
display(left_semi)
```

## **Cross Join**

```python
cross_join = emp_df.crossJoin(dept_df)
cross_join.display()
```

## Select required columns 

```python
left_Join = emp_df.join(dept_df, emp_df.dept_id == dept_df.dept_id, "left")\
    .select(emp_df.emp_name, dept_df.dept_name)

left_Join.display()
```
