```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Aggregations").getOrCreate()

data = [
    (101, "John", "IT", 50000),
    (102, "Alice", "IT", 60000),
    (103, "Bob", "HR", 45000),
    (104, "David", "HR", 55000),
    (105, "Emma", "Finance", 70000),
    (106, "Mike", "Finance", 65000),
    (107, "Sara", "IT", 75000),
    (108, "Tom", "HR", 40000)
]

columns = ["Emp_id","Emp_Name", "Dept", "Salary"]
df = spark.createDataFrame(data, columns)

df.display()
```

### **Import Aggregation Functions**

**1. Count()** -
Management wants total employee count.

```python
from pyspark.sql.functions import *

df.select(count("*").alias("Total_Employees")).display()
```

###2. **sum()** - Total Salary 

```python
total_salary = df.select(
                    sum("Salary").alias("Total_Salary")
                ).display()
```

### **3. Average()** --->> HR wants average employee salary.

```python
avg_salary = df.select(
               avg("salary").alias("avg_salary")
)
avg_salary.display()
```

### **4. Maximum Salary** --> Find highest-paid employee salary.

```python
df.select(
    max("salary").alias("max_salary")
).show()
```

```python
from pyspark.sql.functions import *

max_sal = df.groupBy('emp_name').agg(max('salary').alias('max_salary'))


max_sal.display()
```

### **Minimum Salary** - Find lowest salary.

```python
df.select(
    min("salary").alias("min_salary")
).display()
```

### **6. Multiple Aggregations**

```python
df.select(
    count("*").alias("total_employees"),
    sum("salary").alias("total_salary"),
    avg("salary").alias("avg_salary"),
    max("salary").alias("highest_salary"),
    min("salary").alias("lowest_salary")
).display()
```
